# Gold layer for account_move model

```python
%sql

SELECT * FROM bakery_pos_develop.silver.account_move
```

```python
%sql

CREATE OR REPLACE TEMP VIEW stg_account_move_gold AS
SELECT account_date AS Date, move_type AS Move_type, customer_name AS Customer_name, COUNT(*) AS Record_count
FROM bakery_pos_develop.silver.account_move
GROUP BY account_date, move_type, customer_name
ORDER BY account_date DESC;


SELECT * FROM stg_account_move_gold
```
