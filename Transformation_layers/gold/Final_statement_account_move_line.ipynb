```python
%sql

CREATE OR REPLACE TEMP VIEW stg_account_move_line_gold AS
SELECT account_move_date AS Date, account_name AS Account, SUM(credit) AS Credit, SUM(debit) AS Debit, SUM(balance) AS Balance
FROM bakery_pos_develop.silver.account_move_line
GROUP BY account_move_date,account_name
ORDER BY account_move_date DESC;

SELECT count(*) FROM stg_account_move_line_gold;
```

```python
%sql
CREATE OR REPLACE TEMP VIEW stg_dim_account_move AS
SELECT account_move_id AS id,
move_name AS name,
reference AS reference,
move_type AS type,
move_state AS state
FROM bakery_pos_develop.silver.account_move
WHERE state != 'deleted';

 
```

```python
%sql

CREATE TABLE IF NOT EXISTS bakery_pos_develop.gold.dim_account_move (
    id BIGINT,
    move_id BIGINT NOT NULL,
    name STRING NOT NULL,
    reference STRING,
    type STRING,
    state STRING,
    created_at TIMESTAMP,
    updated_at TIMESTAMP,
    is_current BOOLEAN NOT NULL,
    PRIMARY KEY(id) not ENFORCED
) USING DELTA;
```

```python
%sql

CREATE TABLE IF NOT EXISTS bakery_pos_develop.gold.dim_account (
    id BIGINT,
    account_id INT NOT NULL,
    account_name STRING NOT NULL,
    created_at TIMESTAMP,
    updated_at TIMESTAMP,
    is_current BOOLEAN NOT NULL,
    PRIMARY KEY(id) not ENFORCED
) USING DELTA;
    

CREATE TABLE IF NOT EXISTS bakery_pos_develop.gold.dim_account_move_date (
    id BIGINT,
    move_date DATE,
    year INT,
    month INT,
    day INT,
    PRIMARY KEY(id) NOT ENFORCED
) USING DELTA;


CREATE TABLE IF NOT EXISTS bakery_pos_develop.gold.dim_product (
    id BIGINT,
    product_id INT NOT NULL,
    product_name STRING NOT NULL,
    created_at TIMESTAMP,
    updated_at TIMESTAMP,
    is_current BOOLEAN NOT NULL,
    PRIMARY KEY (id) NOT ENFORCED
) USING DELTA;


CREATE TABLE IF NOT EXISTS bakery_pos_develop.gold.dim_partner (
    id BIGINT,
    partner_id BIGINT NOT NULL,
    partner_name STRING NOT NULL,
    created_at TIMESTAMP,
    updated_at TIMESTAMP,
    is_current BOOLEAN NOT NULL,
    PRIMARY KEY (id) NOT ENFORCED
) USING DELTA;


CREATE TABLE IF NOT EXISTS bakery_pos_develop.gold.dim_journal (
    id BIGINT,
    journal_id INT NOT NULL,
    journal_name STRING,
    created_at TIMESTAMP,
    updated_at TIMESTAMP,
    is_current BOOLEAN NOT NULL,
    PRIMARY KEY (id) NOT ENFORCED

)
```
